## Data dictionary

- `post_url`: Public URL of the featured blog post.
- `published_at_utc`: Publication timestamp supplied by the RSS feed, converted to UTC.
- `collected_at_utc`: UTC timestamp indicating when the RSS feed was retrieved.
- `age_hours`: Difference between collection time and publication time, measured in hours.

Rows without a usable URL or publication timestamp are excluded. Entries with a negative calculated age are also excluded because their publication time is later than the collection time.
