## Data dictionary
- `post_url`: public URL of the post.
- `published_at_utc`: publication timestamp from the RSS feed, converted to UTC.
- `collected_at_utc`: time the feed was retrieved, in UTC.
- `age_hours`: time between publication and collection, in hours.

Rows without a URL or usable publication timestamp, and rows with negative age, are excluded.
